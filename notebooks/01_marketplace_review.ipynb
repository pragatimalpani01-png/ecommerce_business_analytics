# Marketplace performance review

This notebook is a short way to inspect the tables created by the project. Run `python run_project.py --source demo` from the project root first. The default outputs use synthetic demo rows, not the original Olist transactions.

```python
from pathlib import Path
import json
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd()
if not (ROOT / 'outputs' / 'metrics.json').exists():
    ROOT = ROOT.parent
metrics = json.loads((ROOT / 'outputs' / 'metrics.json').read_text())
metrics
```

## Check the order-level table

Each order appears once in this table. The promise gap is positive when the order arrived after its estimated date.

```python
orders = pd.read_csv(ROOT / 'outputs' / 'tables' / 'order_fact.csv', parse_dates=['order_purchase_timestamp'])
print('Rows:', len(orders), '| Unique orders:', orders['order_id'].nunique())
orders[['order_id', 'order_status', 'item_sales_brl', 'freight_brl', 'promise_gap_days', 'review_score']].head(10)
```

## Find categories worth a closer look

The review attached to an order is carried onto its item lines. Treat this as a category screening signal, not a product-specific rating.

```python
categories = pd.read_csv(ROOT / 'outputs' / 'tables' / 'category_performance.csv')
categories[['category', 'orders', 'delivered_item_sales_brl', 'freight_share', 'low_review_share', 'late_share_item_line_proxy']].sort_values('item_sales_brl', ascending=False).head(12)
```
